```python
#Bring in the pandas library
import pandas as pd
```

#SECTION 1: Clean Airbnb Price Data


```python
# Step 1: Load the Airbnb dataset from a CSV file
airbnb = pd.read_csv("/content/airbnb_hw.csv")

# Quick look at the first few rows to see what the raw data looks like
print("Here is a quick peak at the initial table:")
display(airbnb.head())

# Step 2: Clean up the 'Price' column
# The raw price numbers have '$' and ',' signs, which makes Python treat them as plain text.
# We strip those out so Python can recognize them as numbers.
airbnb["Price_cleaned"] = (
    airbnb["Price"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

# Convert the text into real numerical values (numbers)
# 'errors=coerce' safety-net: if something fails to convert, it safely turns into NaN (missing value)
airbnb["Price_cleaned"] = pd.to_numeric(
    airbnb["Price_cleaned"], errors="coerce"
)

# Step 3: Check our clean results
print("\nSide-by-side comparison (Original vs. Cleaned Price):")
display(airbnb[["Price", "Price_cleaned"]].head())

# Count any missing values in the new cleaned column
missing_prices = airbnb["Price_cleaned"].isna().sum()
print(f"Missing prices after cleaning: {missing_prices}")

# Summary statistics (mean, min, max, etc.) for prices
print("\nBasic price summary stats:")
display(airbnb["Price_cleaned"].describe())
```

# SECTION 2: Clean Police Use of Force Data

```python
# Step 1: Load the Minneapolis police dataset
police = pd.read_csv("/content/mn_police_use_of_force.csv")

# See how many missing entries are in 'subject_injury' right out of the box
missing_percent = police["subject_injury"].isna().mean() * 100
print(f"Initial missing injury values: {missing_percent:.2f}%")

# Step 2: Clean the 'subject_injury' column
# We replace empty/NaN spaces and 'False' with 'No', and keep 'True' or 'Yes' as 'Yes'.
police["subject_injury_cleaned"] = police["subject_injury"].fillna("No")
police["subject_injury_cleaned"] = police["subject_injury_cleaned"].replace(
    {"False": "No", "True": "Yes"}
)

# Verify the changes
print("\nCleaned injury value counts:")
print(police["subject_injury_cleaned"].value_counts())

# Step 3: Compare cleaned injury data against the 'force_type' used
print("\nInjury counts grouped by Force Type:")
display(pd.crosstab(police["subject_injury_cleaned"], police["force_type"]))
```

# SECTION 3: Load & Clean Shark Attack Data

```python
# Step 1: Download the Excel file directly from the web using pandas
file_url = "https://sharkattackfile.net/spreadsheets/GSAF5.xls"
sharks = pd.read_excel(file_url)

print("\nInitial Shark Dataset loaded!")
display(sharks.head(3))

# Step 2: Drop completely empty columns
# 'how=all' tells pandas to only delete columns where EVERY single row is empty (NaN)
# 'thresh=1' ensures we drop columns that don't have at least 1 valid entry
sharks_cleaned = sharks.dropna(how="all", axis=1)

# Report back on how many columns remain
print(f"Original column count: {sharks.shape[1]}")
print(f"Column count after dropping empty columns: {sharks_cleaned.shape[1]}")
```
